```python
import pandas as pd
```

```python


# Load raw deliveries data
df = pd.read_csv("../datasets/original_deliveries.csv")

# Count batting appearances
bat_counts = df['batter'].value_counts()

# Count bowling appearances
bowl_counts = df['bowler'].value_counts()

players = set(bat_counts.index).union(set(bowl_counts.index))

player_roles = {}

for player in players:
    bat = bat_counts.get(player, 0)
    bowl = bowl_counts.get(player, 0)

    if bat > 3 * bowl:
        role = "batsman"
    elif bowl > 3 * bat:
        role = "bowler"
    else:
        role = "allrounder"

    player_roles[player] = role

# Save roles
roles_df = pd.DataFrame(
    list(player_roles.items()),
    columns=["player", "role"]
)

roles_df.to_csv("../datasets/player_role.csv", index=False)
```
