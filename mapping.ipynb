```python
import pandas as pd

# Load raw Kaggle data
deliveries = pd.read_csv("../datasets/deliveries.csv")
matches = pd.read_csv("../datasets/matches.csv")

# Aggregate batter performance per match WITH TEAM
player_match = (
    deliveries
    .groupby(["match_id", "batter", "batting_team"])
    .agg(
        runs=("batsman_runs", "sum"),
        balls=("ball", "count")
    )
    .reset_index()
)

# Strike rate
player_match["strike_rate"] = (
    player_match["runs"] / player_match["balls"]
) * 100

# Add venue info
player_match = player_match.merge(
    matches[["id", "venue"]],
    left_on="match_id",
    right_on="id",
    how="left"
)

# Rename for clarity
player_match.rename(columns={
    "batting_team": "team"
}, inplace=True)

# Save mapping dataset
player_match.to_csv("../datasets/mapping_data.csv", index=False)

print("mapping_data.csv created successfully with TEAM column")
```
